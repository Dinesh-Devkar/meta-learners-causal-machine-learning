```python
# ============================================================
# META-LEARNERS USING ECONML
# Dataset: Criteo Uplift Prediction Dataset
#
# Meta-Learners:
# 1. S-Learner
# 2. T-Learner
# 3. X-Learner
# 4. R-Learner  -> implemented using NonParamDML
# 5. DR-Learner
#
# Treatment : treatment
# Outcome   : conversion
# Features  : f0 - f11
#
# visit and exposure are NOT used as model features.
# ============================================================


# ============================================================
# 0. INSTALLATION
# ============================================================

# %pip install -q econml xgboost


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier

from xgboost import XGBRegressor

from econml.metalearners import (
    SLearner,
    TLearner,
    XLearner
)

from econml.dml import NonParamDML
from econml.dr import DRLearner


RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)


# ============================================================
# 2. LOAD CRITEO DATASET
# ============================================================

# Update path according to your dataset location

df = pd.read_parquet('./drive/MyDrive/Untitledfolder/0001.parquet')


print(
    "Dataset shape:",
    df.shape
)

display(
    df.head()
)


# ============================================================
# 3. DEFINE FEATURES
# ============================================================

FEATURES = [
    "f0",
    "f1",
    "f2",
    "f3",
    "f4",
    "f5",
    "f6",
    "f7",
    "f8",
    "f9",
    "f10",
    "f11"
]


TREATMENT_COL = "treatment"

OUTCOME_COL = "conversion"


print("Features:")
print(FEATURES)


print("\nTreatment column:")
print(TREATMENT_COL)


print("\nOutcome column:")
print(OUTCOME_COL)


print(
    "\nColumns intentionally NOT used as features:"
)

print(
    ["visit", "exposure"]
)


# ============================================================
# 4. OPTIONAL SAMPLING
# ============================================================
#
# Criteo is a very large dataset.
#
# Start with 500K observations while developing.
#
# Set:
#
# SAMPLE_SIZE = None
#
# to use the complete dataset.
# ============================================================

SAMPLE_SIZE = 500_000


if SAMPLE_SIZE is not None and len(df) > SAMPLE_SIZE:

    df_model = df.sample(
        n=SAMPLE_SIZE,
        random_state=RANDOM_STATE
    ).copy()

else:

    df_model = df.copy()


print(
    "Modeling dataset shape:",
    df_model.shape
)


# ============================================================
# 5. BASIC DATA CHECKS
# ============================================================

print("\nMissing values:")

print(
    df_model[
        FEATURES
        + [TREATMENT_COL, OUTCOME_COL]
    ].isnull().sum()
)


print("\nTreatment distribution:")

print(
    df_model[
        TREATMENT_COL
    ].value_counts(
        normalize=True
    )
)


print("\nOutcome distribution:")

print(
    df_model[
        OUTCOME_COL
    ].value_counts(
        normalize=True
    )
)


print("\nConversion rate by treatment:")

print(
    df_model
    .groupby(
        TREATMENT_COL
    )[OUTCOME_COL]
    .agg(
        ["mean", "count"]
    )
)


# ============================================================
# 6. CREATE X, T AND Y
# ============================================================

X = (
    df_model[FEATURES]
    .astype("float32")
)


T = (
    df_model[TREATMENT_COL]
    .astype("int8")
)


Y = (
    df_model[OUTCOME_COL]
    .astype("int8")
)


print(
    "\nX shape:",
    X.shape
)

print(
    "Treatment rate:",
    T.mean()
)

print(
    "Conversion rate:",
    Y.mean()
)


# ============================================================
# 7. TRAIN / TEST SPLIT
# ============================================================
#
# Since conversion is rare, stratify using both:
#
# Treatment + Outcome
#
# This helps preserve:
#
# T=0, Y=0
# T=0, Y=1
# T=1, Y=0
# T=1, Y=1
#
# proportions across train/test.
# ============================================================

strata = (
    T.astype(str)
    + "_"
    + Y.astype(str)
)


(
    X_train,
    X_test,
    T_train,
    T_test,
    Y_train,
    Y_test
) = train_test_split(

    X,
    T,
    Y,

    test_size=0.30,

    random_state=RANDOM_STATE,

    stratify=strata
)


# Reset indexes

X_train = (
    X_train
    .reset_index(drop=True)
)

X_test = (
    X_test
    .reset_index(drop=True)
)


T_train = (
    T_train
    .reset_index(drop=True)
)

T_test = (
    T_test
    .reset_index(drop=True)
)


Y_train = (
    Y_train
    .reset_index(drop=True)
)

Y_test = (
    Y_test
    .reset_index(drop=True)
)


print(
    "\nTrain shape:",
    X_train.shape
)

print(
    "Test shape:",
    X_test.shape
)


print(
    "\nTrain treatment rate:",
    T_train.mean()
)

print(
    "Test treatment rate:",
    T_test.mean()
)


print(
    "\nTrain conversion rate:",
    Y_train.mean()
)

print(
    "Test conversion rate:",
    Y_test.mean()
)


# ============================================================
# 8. OBSERVED EXPERIMENTAL ATE
# ============================================================
#
# Because Criteo treatment assignment is randomized:
#
#
# ATE =
#
# E[Y | T=1]
#      -
# E[Y | T=0]
#
#
# gives us a useful experimental benchmark.
# ============================================================

treated_conversion = (
    Y_test[
        T_test == 1
    ].mean()
)


control_conversion = (
    Y_test[
        T_test == 0
    ].mean()
)


observed_ate = (
    treated_conversion
    -
    control_conversion
)


print(
    "\n========================================"
)

print(
    "OBSERVED RANDOMIZED EXPERIMENT"
)

print(
    "========================================"
)


print(
    f"Treated conversion rate : "
    f"{treated_conversion:.6f}"
)


print(
    f"Control conversion rate : "
    f"{control_conversion:.6f}"
)


print(
    f"Observed ATE             : "
    f"{observed_ate:.6f}"
)


# ============================================================
# 9. PROPENSITY / TREATMENT ASSIGNMENT
# ============================================================
#
# Normally:
#
# e(X) = P(T=1 | X)
#
#
# But Criteo comes from a randomized experiment.
#
# Therefore treatment probability should NOT meaningfully
# depend on X.
#
#
# Instead of training something like:
#
# XGBoostClassifier(X -> T)
#
#
# we use:
#
# DummyClassifier(strategy="prior")
#
#
# This learns only the overall treatment probability.
#
# ============================================================

treatment_probability = (
    T_train.mean()
)


print(
    "\nTreatment assignment probability:",
    treatment_probability
)


# Helper function
#
# Create a fresh propensity model whenever needed.

def make_propensity_model():

    return DummyClassifier(
        strategy="prior"
    )


# ============================================================
# 10. COMMON XGBOOST CONFIGURATION
# ============================================================
#
# Same base learner configuration for all estimators as much
# as possible.
#
# This makes the comparison primarily about:
#
# S vs T vs X vs R vs DR
#
# rather than differences in underlying ML algorithms.
# ============================================================

xgb_params = {

    "n_estimators": 200,

    "max_depth": 5,

    "learning_rate": 0.05,

    "subsample": 0.8,

    "colsample_bytree": 0.8,

    "objective": "reg:squarederror",

    "tree_method": "hist",

    "random_state": RANDOM_STATE,

    "n_jobs": -1
}


# Helper function
#
# Every causal component receives a fresh estimator.

def make_xgb():

    return XGBRegressor(
        **xgb_params
    )


# ============================================================
# ============================================================
#
#                    1. S-LEARNER
#
# ============================================================
# ============================================================
#
# Mental model:
#
#
#             X + T
#               |
#               v
#         Single outcome model
#
#
# Same user:
#
# T = 0
#   |
#   v
# Y_hat(0)
#
#
# T = 1
#   |
#   v
# Y_hat(1)
#
#
# CATE:
#
# tau(X) = Y_hat(1) - Y_hat(0)
#
# ============================================================

print(
    "\n\n========================================"
)

print(
    "TRAINING S-LEARNER"
)

print(
    "========================================"
)


s_learner = SLearner(

    overall_model=make_xgb(),

    categories=[0, 1]
)


s_learner.fit(

    Y_train,

    T_train,

    X=X_train
)


cate_s = s_learner.effect(

    X_test,

    T0=0,

    T1=1

).reshape(-1)


print(
    "\nS-Learner CATE summary:"
)


print(
    pd.Series(
        cate_s
    ).describe()
)


print(
    "\nS-Learner Mean Predicted CATE:",
    cate_s.mean()
)


# ============================================================
# ============================================================
#
#                    2. T-LEARNER
#
# ============================================================
# ============================================================
#
# Split observations:
#
#
#               Dataset
#                  |
#          ------------------
#          |                |
#       Control          Treatment
#
#          |                |
#
#       mu_0(X)          mu_1(X)
#
#          |                |
#
#       Y_hat(0)         Y_hat(1)
#
#
#
# CATE:
#
# tau(X) = mu_1(X) - mu_0(X)
#
# ============================================================

print(
    "\n\n========================================"
)

print(
    "TRAINING T-LEARNER"
)

print(
    "========================================"
)


t_learner = TLearner(

    models=[
        make_xgb(),
        make_xgb()
    ],

    categories=[0, 1]
)


t_learner.fit(

    Y_train,

    T_train,

    X=X_train
)


cate_t = t_learner.effect(

    X_test,

    T0=0,

    T1=1

).reshape(-1)


print(
    "\nT-Learner CATE summary:"
)


print(
    pd.Series(
        cate_t
    ).describe()
)


print(
    "\nT-Learner Mean Predicted CATE:",
    cate_t.mean()
)


# ============================================================
# ============================================================
#
#                    3. X-LEARNER
#
# ============================================================
# ============================================================
#
# STEP 1
#
# Learn separate outcome models:
#
# mu_0(X)
# mu_1(X)
#
#
# STEP 2
#
# Construct pseudo effects.
#
#
# Treated observations:
#
# D1 = Y - mu_0(X)
#
#
# Control observations:
#
# D0 = mu_1(X) - Y
#
#
# STEP 3
#
# Learn:
#
# tau_0(X)
# tau_1(X)
#
#
# STEP 4
#
# Combine using propensity:
#
# e(X) = P(T=1 | X)
#
#
# Since treatment is randomized,
#
# DummyClassifier(strategy="prior")
#
# represents our constant propensity.
#
# ============================================================

print(
    "\n\n========================================"
)

print(
    "TRAINING X-LEARNER"
)

print(
    "========================================"
)


x_learner = XLearner(

    # Outcome models:
    #
    # model for control
    # model for treated

    models=[
        make_xgb(),
        make_xgb()
    ],


    # Models for pseudo treatment effects

    cate_models=[
        make_xgb(),
        make_xgb()
    ],


    # Randomized treatment propensity

    propensity_model=make_propensity_model(),


    categories=[0, 1]
)


x_learner.fit(

    Y_train,

    T_train,

    X=X_train
)


cate_x = x_learner.effect(

    X_test,

    T0=0,

    T1=1

).reshape(-1)


print(
    "\nX-Learner CATE summary:"
)


print(
    pd.Series(
        cate_x
    ).describe()
)


print(
    "\nX-Learner Mean Predicted CATE:",
    cate_x.mean()
)


# ============================================================
# ============================================================
#
#                    4. R-LEARNER
#
#                  via NonParamDML
#
# ============================================================
# ============================================================
#
# IMPORTANT
# ---------
#
# EconML does not expose a public class named:
#
# RLearner
#
#
# EconML's DML estimators implement the R-Learner /
# residual-on-residual framework.
#
#
# We use NonParamDML because we want:
#
# nonlinear CATE tau(X)
#
# using XGBoost as the final model.
#
#
#
# STEP 1
#
# Outcome nuisance model:
#
# m(X) = E[Y | X]
#
#
# STEP 2
#
# Propensity:
#
# e(X) = E[T | X]
#
#
# STEP 3
#
# Residualize:
#
# Y_res = Y - m(X)
#
# T_res = T - e(X)
#
#
# STEP 4
#
# Learn heterogeneous treatment effect:
#
# Y_res ≈ T_res * tau(X)
#
# ============================================================

print(
    "\n\n========================================"
)

print(
    "TRAINING R-LEARNER / NonParamDML"
)

print(
    "========================================"
)


r_learner = NonParamDML(

    # Outcome nuisance model
    model_y=make_xgb(),


    # Treatment nuisance model
    #
    # Constant propensity because Criteo is randomized.

    model_t=make_propensity_model(),


    # Final heterogeneous CATE model

    model_final=make_xgb(),


    # Treatment is binary/categorical

    discrete_treatment=True,


    # We keep Y as numeric 0/1 and estimate E[Y|X]
    # with a regressor.

    discrete_outcome=False,


    categories=[0, 1],


    # Cross-fitting

    cv=5,


    random_state=RANDOM_STATE
)


r_learner.fit(

    Y_train,

    T_train,

    X=X_train
)


cate_r = r_learner.effect(

    X_test,

    T0=0,

    T1=1

).reshape(-1)


print(
    "\nR-Learner CATE summary:"
)


print(
    pd.Series(
        cate_r
    ).describe()
)


print(
    "\nR-Learner Mean Predicted CATE:",
    cate_r.mean()
)


# ============================================================
# ============================================================
#
#                    5. DR-LEARNER
#
# ============================================================
# ============================================================
#
# DR-Learner combines:
#
#
# Outcome model
#
# mu_t(X) = E[Y | X, T=t]
#
#
#              +
#
#
# Propensity model
#
# e(X) = P(T | X)
#
#
#              |
#              v
#
#     Doubly Robust pseudo-outcome
#
#              |
#              v
#
#       Final CATE model
#
#           tau(X)
#
#
# The attractive property is that the pseudo-outcome uses
# information from BOTH:
#
# outcome regression
#
# and
#
# propensity estimation.
#
# ============================================================

print(
    "\n\n========================================"
)

print(
    "TRAINING DR-LEARNER"
)

print(
    "========================================"
)


dr_learner = DRLearner(

    # Treatment / propensity model
    #
    # Constant treatment probability because randomized.

    model_propensity=make_propensity_model(),


    # Outcome regression

    model_regression=make_xgb(),


    # Final heterogeneous effect model

    model_final=make_xgb(),


    discrete_outcome=False,


    categories=[0, 1],


    cv=5,


    random_state=RANDOM_STATE
)


dr_learner.fit(

    Y_train,

    T_train,

    X=X_train
)


cate_dr = dr_learner.effect(

    X_test,

    T0=0,

    T1=1

).reshape(-1)


print(
    "\nDR-Learner CATE summary:"
)


print(
    pd.Series(
        cate_dr
    ).describe()
)


print(
    "\nDR-Learner Mean Predicted CATE:",
    cate_dr.mean()
)


# ============================================================
# 11. CREATE RESULTS DATAFRAME
# ============================================================

results = pd.DataFrame({

    "treatment":
        T_test.to_numpy(),

    "conversion":
        Y_test.to_numpy(),

    "S_Learner":
        cate_s,

    "T_Learner":
        cate_t,

    "X_Learner":
        cate_x,

    "R_Learner":
        cate_r,

    "DR_Learner":
        cate_dr
})


print(
    "\n\n========================================"
)

print(
    "PREDICTION RESULTS"
)

print(
    "========================================"
)


display(
    results.head()
)


# ============================================================
# 12. COMPARE MEAN PREDICTED CATE
# ============================================================

comparison = pd.DataFrame({

    "Model": [

        "S-Learner",

        "T-Learner",

        "X-Learner",

        "R-Learner",

        "DR-Learner"
    ],


    "Mean Predicted CATE": [

        cate_s.mean(),

        cate_t.mean(),

        cate_x.mean(),

        cate_r.mean(),

        cate_dr.mean()
    ]
})


comparison[
    "Observed ATE"
] = observed_ate


comparison[
    "Difference From Observed ATE"
] = (

    comparison[
        "Mean Predicted CATE"
    ]

    -

    comparison[
        "Observed ATE"
    ]
)


print(
    "\n\n========================================"
)

print(
    "MEAN CATE COMPARISON"
)

print(
    "========================================"
)


display(
    comparison
)


# ============================================================
# 13. CATE DISTRIBUTION
# ============================================================

cate_columns = [

    "S_Learner",

    "T_Learner",

    "X_Learner",

    "R_Learner",

    "DR_Learner"
]


cate_summary = (

    results[
        cate_columns
    ]

    .describe()

    .T
)


print(
    "\n\n========================================"
)

print(
    "CATE DISTRIBUTION"
)

print(
    "========================================"
)


display(
    cate_summary
)


# ============================================================
# 14. CUSTOM QINI / UPLIFT EVALUATION
# ============================================================
#
# EconML focuses primarily on causal effect estimation and does
# not provide the same Qini convenience functions used by
# CausalML.
#
# To keep this notebook independent from CausalML, below we
# implement the evaluation directly.
#
#
# For each model:
#
# 1. Rank users by predicted CATE
#
# 2. Move from highest predicted responder to lowest
#
# 3. Calculate cumulative incremental conversions
#
#
# Qini(k) =
#
# cumulative treated conversions
#
# -
#
# cumulative control conversions
# * cumulative treated count / cumulative control count
#
# ============================================================


def calculate_qini_curve(
    data,
    score_col,
    treatment_col="treatment",
    outcome_col="conversion"
):

    temp = (
        data[
            [
                treatment_col,
                outcome_col,
                score_col
            ]
        ]
        .sort_values(
            score_col,
            ascending=False
        )
        .reset_index(drop=True)
    )


    treatment = (
        temp[treatment_col]
        .to_numpy()
    )


    outcome = (
        temp[outcome_col]
        .to_numpy()
    )


    # Cumulative number of treated observations

    cum_treated = np.cumsum(
        treatment == 1
    )


    # Cumulative number of controls

    cum_control = np.cumsum(
        treatment == 0
    )


    # Cumulative outcome among treated

    cum_y_treated = np.cumsum(

        outcome
        * (treatment == 1)
    )


    # Cumulative outcome among controls

    cum_y_control = np.cumsum(

        outcome
        * (treatment == 0)
    )


    # Expected control outcomes scaled to size of
    # the treated population

    expected_control_for_treated = np.divide(

        cum_y_control
        * cum_treated,

        cum_control,

        out=np.zeros_like(
            cum_y_control,
            dtype=float
        ),

        where=(
            cum_control != 0
        )
    )


    # Incremental conversions

    qini = (

        cum_y_treated

        -

        expected_control_for_treated
    )


    population_fraction = (

        np.arange(
            1,
            len(temp) + 1
        )

        /

        len(temp)
    )


    return (
        population_fraction,
        qini
    )


# ============================================================
# 15. QINI COEFFICIENT
# ============================================================


def calculate_qini_score(
    data,
    score_col
):

    x, qini = calculate_qini_curve(
        data,
        score_col
    )


    # Add origin

    x = np.insert(
        x,
        0,
        0
    )


    qini = np.insert(
        qini,
        0,
        0
    )


    # Random targeting baseline

    random_baseline = (

        x
        * qini[-1]
    )


    # Area between model Qini curve and random baseline

    qini_score = np.trapz(

        qini
        -
        random_baseline,

        x
    )


    return qini_score


# ============================================================
# 16. CALCULATE QINI SCORES
# ============================================================

qini_results = []


for model_name in cate_columns:

    score = calculate_qini_score(

        results,

        model_name
    )


    qini_results.append({

        "Model":
            model_name,

        "Qini Score":
            score
    })


qini_results = pd.DataFrame(
    qini_results
)


print(
    "\n\n========================================"
)

print(
    "QINI SCORES"
)

print(
    "========================================"
)


display(

    qini_results.sort_values(

        "Qini Score",

        ascending=False
    )
)


# ============================================================
# 17. PLOT QINI CURVES
# ============================================================

plt.figure(
    figsize=(12, 8)
)


for model_name in cate_columns:

    x, qini = calculate_qini_curve(

        results,

        model_name
    )


    # Add origin

    x_plot = np.insert(
        x,
        0,
        0
    )


    qini_plot = np.insert(
        qini,
        0,
        0
    )


    plt.plot(

        x_plot,

        qini_plot,

        label=model_name
    )


# Random baseline
#
# Use overall experiment incremental outcome.

random_x = np.array([
    0,
    1
])


# Calculate final Qini based on complete test population

_, overall_qini = calculate_qini_curve(

    results,

    cate_columns[0]
)


random_y = np.array([
    0,
    overall_qini[-1]
])


plt.plot(

    random_x,

    random_y,

    linestyle="--",

    label="Random"
)


plt.xlabel(
    "Fraction of Population Targeted"
)


plt.ylabel(
    "Cumulative Incremental Conversions"
)


plt.title(
    "Qini Curves - EconML Meta-Learners"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.show()


# ============================================================
# 18. CUMULATIVE UPLIFT CURVE / AUUC
# ============================================================


def calculate_uplift_curve(
    data,
    score_col
):

    temp = (

        data[
            [
                "treatment",
                "conversion",
                score_col
            ]
        ]

        .sort_values(
            score_col,
            ascending=False
        )

        .reset_index(drop=True)
    )


    treatment = (
        temp["treatment"]
        .to_numpy()
    )


    outcome = (
        temp["conversion"]
        .to_numpy()
    )


    cum_treated = np.cumsum(
        treatment == 1
    )


    cum_control = np.cumsum(
        treatment == 0
    )


    cum_y_treated = np.cumsum(

        outcome
        * (treatment == 1)
    )


    cum_y_control = np.cumsum(

        outcome
        * (treatment == 0)
    )


    treated_rate = np.divide(

        cum_y_treated,

        cum_treated,

        out=np.zeros_like(
            cum_y_treated,
            dtype=float
        ),

        where=(
            cum_treated != 0
        )
    )


    control_rate = np.divide(

        cum_y_control,

        cum_control,

        out=np.zeros_like(
            cum_y_control,
            dtype=float
        ),

        where=(
            cum_control != 0
        )
    )


    uplift = (

        treated_rate

        -

        control_rate
    )


    population_fraction = (

        np.arange(
            1,
            len(temp) + 1
        )

        /

        len(temp)
    )


    return (
        population_fraction,
        uplift
    )


# ============================================================
# 19. AUUC SCORES
# ============================================================


def calculate_auuc(
    data,
    score_col
):

    x, uplift = calculate_uplift_curve(

        data,

        score_col
    )


    x = np.insert(
        x,
        0,
        0
    )


    uplift = np.insert(
        uplift,
        0,
        0
    )


    return np.trapz(
        uplift,
        x
    )


auuc_results = []


for model_name in cate_columns:

    auuc = calculate_auuc(

        results,

        model_name
    )


    auuc_results.append({

        "Model":
            model_name,

        "AUUC":
            auuc
    })


auuc_results = pd.DataFrame(
    auuc_results
)


print(
    "\n\n========================================"
)

print(
    "AUUC SCORES"
)

print(
    "========================================"
)


display(

    auuc_results.sort_values(

        "AUUC",

        ascending=False
    )
)


# ============================================================
# 20. TOP 20% TARGETING VALIDATION
# ============================================================
#
# Business interpretation:
#
#
# Model predicts:
#
# "These users should benefit most from treatment."
#
#
# We then verify using the randomized experiment:
#
# E[Y | T=1, Top 20%]
#
#       -
#
# E[Y | T=0, Top 20%]
#
#
# Is this larger than the population ATE?
#
# ============================================================

top20_results = []


for model_name in cate_columns:

    temp = (

        results[
            [
                "treatment",
                "conversion",
                model_name
            ]
        ]

        .sort_values(
            model_name,
            ascending=False
        )

        .reset_index(drop=True)
    )


    top_20_n = int(
        len(temp) * 0.20
    )


    top_20 = temp.iloc[
        :top_20_n
    ]


    treated_rate = (

        top_20.loc[

            top_20[
                "treatment"
            ] == 1,

            "conversion"
        ]

        .mean()
    )


    control_rate = (

        top_20.loc[

            top_20[
                "treatment"
            ] == 0,

            "conversion"
        ]

        .mean()
    )


    top20_uplift = (

        treated_rate

        -

        control_rate
    )


    top20_results.append({

        "Model":
            model_name,

        "Top 20% Treated Conversion":
            treated_rate,

        "Top 20% Control Conversion":
            control_rate,

        "Top 20% Observed Uplift":
            top20_uplift,

        "Overall Observed ATE":
            observed_ate
    })


top20_results = pd.DataFrame(
    top20_results
)


print(
    "\n\n========================================"
)

print(
    "TOP 20% TARGETING VALIDATION"
)

print(
    "========================================"
)


display(

    top20_results.sort_values(

        "Top 20% Observed Uplift",

        ascending=False
    )
)


# ============================================================
# 21. BOTTOM 20% TARGETING VALIDATION
# ============================================================

bottom20_results = []


for model_name in cate_columns:

    temp = (

        results[
            [
                "treatment",
                "conversion",
                model_name
            ]
        ]

        .sort_values(
            model_name,
            ascending=False
        )

        .reset_index(drop=True)
    )


    bottom_20_n = int(
        len(temp) * 0.20
    )


    bottom_20 = temp.iloc[
        -bottom_20_n:
    ]


    treated_rate = (

        bottom_20.loc[

            bottom_20[
                "treatment"
            ] == 1,

            "conversion"
        ]

        .mean()
    )


    control_rate = (

        bottom_20.loc[

            bottom_20[
                "treatment"
            ] == 0,

            "conversion"
        ]

        .mean()
    )


    bottom20_uplift = (

        treated_rate

        -

        control_rate
    )


    bottom20_results.append({

        "Model":
            model_name,

        "Bottom 20% Treated Conversion":
            treated_rate,

        "Bottom 20% Control Conversion":
            control_rate,

        "Bottom 20% Observed Uplift":
            bottom20_uplift,

        "Overall Observed ATE":
            observed_ate
    })


bottom20_results = pd.DataFrame(
    bottom20_results
)


print(
    "\n\n========================================"
)

print(
    "BOTTOM 20% TARGETING VALIDATION"
)

print(
    "========================================"
)


display(
    bottom20_results
)


# ============================================================
# 22. TOP 20% VS BOTTOM 20%
# ============================================================
#
# A useful heterogeneous treatment effect model should
# ideally produce:
#
#
# Top 20% observed uplift
#
#       >
#
# Bottom 20% observed uplift
#
#
# The larger the separation, the better the model may be at
# ranking treatment responders.
#
# ============================================================

heterogeneity_comparison = (

    top20_results[
        [
            "Model",
            "Top 20% Observed Uplift"
        ]
    ]

    .merge(

        bottom20_results[
            [
                "Model",
                "Bottom 20% Observed Uplift"
            ]
        ],

        on="Model"
    )
)


heterogeneity_comparison[
    "Top-Bottom Difference"
] = (

    heterogeneity_comparison[
        "Top 20% Observed Uplift"
    ]

    -

    heterogeneity_comparison[
        "Bottom 20% Observed Uplift"
    ]
)


print(
    "\n\n========================================"
)

print(
    "TOP 20% VS BOTTOM 20%"
)

print(
    "========================================"
)


display(

    heterogeneity_comparison.sort_values(

        "Top-Bottom Difference",

        ascending=False
    )
)


# ============================================================
# 23. FINAL MODEL COMPARISON
# ============================================================

final_comparison = (

    comparison

    .merge(

        qini_results,

        left_on=(
            comparison[
                "Model"
            ]
            .str.replace(
                "-",
                "_"
            )
        ),

        right_on="Model",

        how="left"
    )
)


# Because merge above generates duplicate model labels,
# construct the final table explicitly for clarity.

final_comparison = pd.DataFrame({

    "Model": [
        "S-Learner",
        "T-Learner",
        "X-Learner",
        "R-Learner",
        "DR-Learner"
    ],

    "Mean Predicted CATE": [
        cate_s.mean(),
        cate_t.mean(),
        cate_x.mean(),
        cate_r.mean(),
        cate_dr.mean()
    ],

    "Observed ATE": [
        observed_ate
    ] * 5
})


# Mapping between display names and results columns

model_mapping = {

    "S-Learner":
        "S_Learner",

    "T-Learner":
        "T_Learner",

    "X-Learner":
        "X_Learner",

    "R-Learner":
        "R_Learner",

    "DR-Learner":
        "DR_Learner"
}


qini_dict = (

    qini_results

    .set_index(
        "Model"
    )["Qini Score"]

    .to_dict()
)


auuc_dict = (

    auuc_results

    .set_index(
        "Model"
    )["AUUC"]

    .to_dict()
)


top20_dict = (

    top20_results

    .set_index(
        "Model"
    )[
        "Top 20% Observed Uplift"
    ]

    .to_dict()
)


bottom20_dict = (

    bottom20_results

    .set_index(
        "Model"
    )[
        "Bottom 20% Observed Uplift"
    ]

    .to_dict()
)


final_comparison[
    "Qini Score"
] = (

    final_comparison[
        "Model"
    ]

    .map(model_mapping)

    .map(qini_dict)
)


final_comparison[
    "AUUC"
] = (

    final_comparison[
        "Model"
    ]

    .map(model_mapping)

    .map(auuc_dict)
)


final_comparison[
    "Top 20% Observed Uplift"
] = (

    final_comparison[
        "Model"
    ]

    .map(model_mapping)

    .map(top20_dict)
)


final_comparison[
    "Bottom 20% Observed Uplift"
] = (

    final_comparison[
        "Model"
    ]

    .map(model_mapping)

    .map(bottom20_dict)
)


final_comparison[
    "Top-Bottom Difference"
] = (

    final_comparison[
        "Top 20% Observed Uplift"
    ]

    -

    final_comparison[
        "Bottom 20% Observed Uplift"
    ]
)


print(
    "\n\n========================================"
)

print(
    "FINAL ECONML META-LEARNER COMPARISON"
)

print(
    "========================================"
)


display(

    final_comparison.sort_values(

        "Qini Score",

        ascending=False
    )
)


# ============================================================
# END
# ============================================================
```
